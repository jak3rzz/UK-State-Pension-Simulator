```python
import matplotlib.pyplot as plt
import numpy as np

#Version 1 - Current Triple Lock

pension_index = float(input("Choose a starting pension index: "))
while pension_index <= 0:
  pension_index = float(input("Choose a new starting pension index. The number must be positive: "))
average_earnings_index = 100

inflation = float(input("Enter UK inflation rate (%): ")) / 100
if inflation <= -1:
  raise ValueError('Invalid economic scenario')
average_earnings_growth = float(input("Enter UK average earnings growth rate (%): ")) / 100
if average_earnings_growth <= -1:
  raise ValueError('Invalid economic scenario')
pension_rise_floor = 0.025

simulation_horizon_in_years = int(input("Enter a simulation horizon in years: "))
while simulation_horizon_in_years < 1:
  simulation_horizon_in_years = int(input(" The simulation horizon must be at least one year. Enter a new value: "))
years = list(range(0, simulation_horizon_in_years + 1))

pension_history = [pension_index]
average_earnings_history = [average_earnings_index]
relative_value_history = [pension_index / average_earnings_index]

for year in range(1, simulation_horizon_in_years + 1):
  pension_index = pension_index * (1 + max(inflation, average_earnings_growth, pension_rise_floor))
  average_earnings_index = average_earnings_index * (1 + average_earnings_growth)
  relative_value = pension_index / average_earnings_index
  pension_history.append(pension_index)
  average_earnings_history.append(average_earnings_index)
  relative_value_history.append(relative_value)
  #print(year, f"{pension_index:.2f}", f"{average_earnings_index:.2f}", f"{relative_value:.2f}")

plt.plot(years, pension_history, label='State Pension')
plt.plot(years, average_earnings_history, label='Average Earnings')
plt.xlim(left=0)
plt.xlabel('Year')
plt.ylabel('Index')
plt.title('Path of UK Triple Lock Pension vs Average Earnings')
plt.legend()
plt.show()

plt.plot(years, relative_value_history)
plt.xlim(left=0)
plt.axhline(y=1, color='red')
plt.xlabel('Year')
plt.ylabel('Relative Pension Value')
plt.title('Relative Growth of UK State Pension vs Average Earnings')
plt.show()



```
